# Exploratory Data Analysis (EDA) Report

## Objective

The objective of this project was to perform Exploratory Data Analysis (EDA) on the dataset to identify patterns, trends, relationships, and key factors influencing content characteristics such as genre, ratings, popularity, runtime, and release trends.

## Data Overview

The analysis was conducted using two datasets:

* **raw_titles.csv** containing information about movies and TV shows.
* **raw_credits.csv** containing cast and crew details.

The titles dataset consists of **5,806 records**, while the credits dataset contains **77,213 records**.

Data exploration, cleaning, visualization, and correlation analysis were performed to gain meaningful insights from the data.

## Key Findings

### 1. Content Distribution

The dataset contains both movies and TV shows.

* Movies: **3,759**
* TV Shows: **2,047**

Movies account for approximately **65%** of the dataset, making them the dominant content type.

### 2. Genre Analysis

Drama emerged as the most common genre in the dataset.

Top genres include:

1. Drama – 2,901 titles
2. Comedy – 2,269 titles
3. Thriller – 1,178 titles
4. Action – 1,053 titles
5. Romance – 958 titles

This indicates a strong preference for drama and comedy-based content.

### 3. Release Year Trends

Content production increased significantly in recent years.

The highest number of titles were released in **2019**, with **848 titles**, suggesting rapid growth in content creation and distribution during that period.

### 4. IMDb Ratings Analysis

IMDb ratings were available for most titles.

* Average IMDb Score: **6.53**
* Median IMDb Score: **6.6**
* Highest IMDb Score: **9.6**
* Lowest IMDb Score: **1.5**

Most titles received ratings between **5.8 and 7.4**, indicating generally positive audience reception.

### 5. Runtime Analysis

The runtime analysis revealed that most content falls within a standard viewing duration range. A few outliers were identified, representing exceptionally long productions compared to the majority of titles.

### 6. Popularity and Audience Engagement

A positive relationship was observed between IMDb votes and IMDb scores.

The correlation coefficient between IMDb votes and IMDb scores was approximately **0.19**, indicating that titles receiving more audience attention generally tend to achieve higher ratings.

### 7. Cast Analysis

Several actors appeared repeatedly across multiple productions.

The most frequently occurring actors include:

* Shah Rukh Khan – 30 appearances
* Anupam Kher – 25 appearances
* Kareena Kapoor Khan – 25 appearances
* Boman Irani – 25 appearances
* Takahiro Sakurai – 22 appearances

This suggests that a small group of actors contribute significantly across numerous titles.

### 8. Correlation Analysis

Correlation analysis was conducted on numerical variables including release year, runtime, IMDb score, and IMDb votes.

Key observations:

* IMDb Votes and IMDb Score showed the strongest positive relationship (**0.19**).
* Runtime and IMDb Votes showed a weak positive relationship (**0.14**).
* Release Year and IMDb Votes showed a weak negative relationship (**-0.20**).

Overall, audience popularity appears to have a stronger influence on ratings than runtime or release year.

## Conclusion

The exploratory analysis revealed several meaningful trends within the dataset. Movies dominate the content library, while drama and comedy are the most prevalent genres. Content production peaked in 2019, reflecting substantial industry growth. IMDb ratings indicate generally favorable audience reception, and titles with higher engagement tend to receive better ratings. Additionally, a small number of actors appear frequently across productions, highlighting their prominence within the entertainment industry.

The analysis successfully identified patterns, relationships, and trends that can support future content strategy, recommendation systems, and audience-focused decision-making.
